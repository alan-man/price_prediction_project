<h1>removing columns and expanding to binary columns</h1>

```python

cols_tokeep = [
    "listing_id", "name", "host_since", "host_is_superhost", "latitude", "longitude",
    "host_total_listings_count", "host_has_profile_pic", "host_identity_verified",
    "room_type", "accommodates", "bedrooms", "amenities", "price",
    "minimum_nights", "maximum_nights", "review_scores_rating",
    "review_scores_cleanliness", "review_scores_checkin",
    "review_scores_communication", "review_scores_location",
    "review_scores_value", "instant_bookable"
]

df1 = paris_df[cols_tokeep].copy()


# parsing amenities
df1["amenities_list"] = df1["amenities"].apply(lambda x: ast.literal_eval(x) if isinstance(x, str) and x.startswith("[") else [])

# all the unique amenities
all_amenities = sorted(set(a for sublist in df1["amenities_list"] for a in sublist))

print(f"Unique amenities: {len(all_amenities)}")

print(all_amenities[:15])

# total columns
num_base_features = len(cols_tokeep) - 1  # minus "amenities"
num_amenities_features = len(all_amenities)
total_columns = num_base_features + num_amenities_features

print(f"Base columns: {num_base_features}")
print(f"Amenities columns: {num_amenities_features}")
print(f"Total expected columns: {total_columns}")
```

<h1>there are too many unique amenities so we take the most used</h1>

```python

# Parse amenities
df1["amenities_list"] = df1["amenities"].apply(
    lambda x: [a.strip().lower() for a in ast.literal_eval(x)] 
    if isinstance(x, str) and x.startswith("[") else []
)

# count the freq of each amenity
amenity_counts = Counter(a for sublist in df1["amenities_list"] for a in sublist)
n_listings = len(df1)

# have only amenities used in at least 1% of listings
threshold = 0.01 * n_listings
common_amenities = {a for a, c in amenity_counts.items() if c >= threshold}

print(f"Unique amenities: {len(amenity_counts)}")
print(f"Frequent amenities (>1%): {len(common_amenities)}")

# binary expension
for a in common_amenities:
    clean_col = "Has_" + a.replace(" ", "_").replace("-", "_").replace("/", "_")
    df1[clean_col] = df1["amenities_list"].apply(lambda x: int(a in x))


final_amenities_cols = [c for c in df1.columns if c.startswith("Has_")]
print(f"Final amenity columns: {len(final_amenities_cols)}")
```
