```python
import bs4
import requests
from bs4 import BeautifulSoup

import lxml

import pandas
import urllib

from urllib import request


url='https://www.seloger.com/classified-search?distributionTypes=Buy&estateTypes=House,Apartment&locations=AD08FR31096'
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers)
page = BeautifulSoup(response.content, "html.parser")
print(page)

```

```python
jo = "https://www.seloger.com/classified-search?distributionTypes=Buy&estateTypes=House,Apartment&locations=AD08FR31096&projectTypes=Resale"
req = request.Request(jo, headers={"User-Agent": "Mozilla/5.0"})

request_text = request.urlopen(req).read()
print(request_text[:1000])
#page = bs4.BeautifulSoup(request_text, "lxml")
page = bs4.BeautifulSoup(request_text, "html.parser")

print(page.find("title"))
```
